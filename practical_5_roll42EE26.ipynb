```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
```

```python
file = '/content/drive/MyDrive/B.TECH/5th Sem/MDM Dataset/wireless_signal_quality_dataset.xlsx - Sheet1.csv'
```

```python
file = '/content/wireless_signal_quality_dataset.xlsx - Sheet1.csv'
data = pd.read_csv(file)
print("first five Records")
print(data.head())
print("\nDataset Shape:", data.shape)
```

```python
data.isnull().sum()
```

```python
X = data.drop("Quality", axis=1)
y = data["Quality"]
```

```python
encoder = LabelEncoder()
y = encoder.fit_transform(y)
print("\nDataset Labels:")
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)
```

```python
rf = RandomForestClassifier(n_estimators=100, max_depth=None, random_state=42)
rf.fit(x_train, y_train)
```

```python
y_pread = rf.predict(x_test)
```

```python
ccuracy = accuracy_score(y_test, y_pread)
print("\nAccuracy = {:.2f}%".format(ccuracy*100))
```

```python
print("\nClassification_Report:")
print(classification_report(y_test, y_pread,
                            target_names=encoder.classes_
                            ))
```

```python
cn = confusion_matrix(y_test, y_pread)
print("\nConfusion Matrix:")
print(cn)
```

```python
importance = pd.Series(
    rf.feature_importances_,
    index=X.columns
)
importance = importance.sort_values(ascending=False)
print("\nFeature Importance:")
print(importance
)
```

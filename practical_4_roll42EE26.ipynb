```python
import pandas as pd
import numpy as np
```

```python
import pandas as pd

# step 1 Load dataset
data = pd.read_csv("/content/wireless_signal_quality_dataset.xlsx - Sheet1.csv", sep=',')

print("First Five Records")
print(data.head())

print("\nDataset Shape :", data.shape)
print("\nDataset Info:")
data.info()

#defined Features and target
X = data.drop("Quality", axis=1)
y = data["Quality"]
```

```python
from sklearn.preprocessing import LabelEncoder

# Step 3 : encode traget lables

encoder = LabelEncoder()
y = encoder.fit_transform(y)

print("\nclass Labels:")
print(encoder.classes_)
```

class Labels:
['Excellent' 'Fair' 'Good' 'Poor']

```python
from sklearn.model_selection import train_test_split

# step 4 Splite Dataset

X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)
```

```python
from sklearn.tree import DecisionTreeClassifier

#train decision tree

dt = DecisionTreeClassifier(
    criterion="gini",
    max_depth=5,
    random_state=42
)

dt.fit(X_train, Y_train)
```

DecisionTreeClassifier(max_depth=5, random_state=42)

```python
# step prediction

y_pred = dt.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

#Step 7 Accuracy

accuracy = accuracy_score(Y_test, y_pred)
print("\nAccuracy : {:.2f}%".format(accuracy * 100))
```

```python
from sklearn.metrics import confusion_matrix

#step 8 confution matrix

cm = confusion_matrix(Y_test, y_pred)
print("\nConfusion Matrix :")
print(cm)
```

```python
from sklearn.metrics import classification_report

#step 9  calssification report
print("\nClassification Report")
print(classification_report(
    Y_test,
    y_pred,
    target_names=encoder.classes_
))
```

```python
#step 10 Feature Importance

importance = pd.Series(
    dt.feature_importances_,
    index=X.columns
)

importance = importance.sort_values(ascending=False)

print("\nFeature Importance")
print(importance
)
```

```python
import matplotlib.pyplot as plt

# step 11 plote feature importance
plt.figure(figsize=(10,5))
importance.plot(kind="bar")
plt.title("Feature Importance - Decision Tree")
plt.xlabel("Features")
plt.ylabel("Importance")
plt.tight_layout()
```

```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

#step 12 Visualise Decision Tree

plt.figure(figsize=(18,10))

plot_tree(
    dt,
    feature_names=X.columns,
    class_names=encoder.classes_,
    filled=True,
    rounded=True,
    fontsize=8
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
```

```python
file = '/content/drive/MyDrive/B.TECH/5th Sem/MDM Dataset/wireless_signal_kmeans_dataset(T7).csv'
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
data = pd.read_csv(file)
```

```python
if len(data.columns) == 1:

    print("CSV format issue detected. Repairing...")

    data = pd.read_csv(
        file,
        sep=",",
        engine="python"
    )
```

```python
if len(data.columns) == 1:

    data = data.iloc[:,0].str.split(
        ",",
        expand=True
    )
```

```python
 data.columns = [
        "RSSI",
        "RSRP",
        "RSRQ",
        "SNR",
        "Latency",
        "Packet_Loss"
    ]
```

```python
data.columns = data.columns.str.strip()
```

```python
print("\nDataset Columns:")
print(data.columns)
```

```python
print("\nDataset Preview:")
print(data.head())
```

```python
data = data.dropna()
```

```python
for col in data.columns:

    data[col] = pd.to_numeric(
        data[col],
        errors="coerce"
    )
```

```python
data.dropna(inplace=True)
```

```python
print("\nClean Dataset:")
print(data.head())
```

```python
X = data[
    [
        "RSSI",
        "RSRP",
        "RSRQ",
        "SNR",
        "Latency",
        "Packet_Loss"
    ]
]
```

```python
scaler = StandardScaler()
```

```python
X_scaled = scaler.fit_transform(
    X
)
```

```python
kmeans = KMeans(

    n_clusters=3,

    random_state=42,

    n_init=10

)
```

```python
data["Cluster"] = kmeans.fit_predict(
    X_scaled
)
```

```python
print("\nCluster Count:")
print(
    data["Cluster"].value_counts()
)
```

```python
cluster_strength = data.groupby(
    "Cluster"
)["RSSI"].mean()
```

```python
print("\nAverage RSSI:")
print(cluster_strength)
```

```python
sorted_cluster = cluster_strength.sort_values(
    ascending=False
)
```

```python
quality = {

    sorted_cluster.index[0]: "Good",

    sorted_cluster.index[1]: "Medium",

    sorted_cluster.index[2]: "Poor"

}
```

```python
data["Signal_Quality"] = data["Cluster"].map(
    quality
)
```

```python
print("\nSignal Quality Groups:")
print(
    data[
        [
            "RSSI",
            "Cluster",
            "Signal_Quality"
        ]
    ]
)
```

```python
pca = PCA(
    n_components=2
)

X_pca = pca.fit_transform(
    X_scaled
)
```

```python
plt.figure(figsize=(8,6))
```

```python
sns.scatterplot(

    x=X_pca[:,0],

    y=X_pca[:,1],

    hue=data["Signal_Quality"],

    palette="Set1",

    s=100

)
```

```python
plt.title(
    "K-Means Clustering of Telecom Signal Quality"
)
```

```python
plt.xlabel(
    "Principal Component 1"
)
```

```python
plt.ylabel(
    "Principal Component 2"
)
```

```python
plt.show()
```

```python
centers = scaler.inverse_transform(
    kmeans.cluster_centers_
)
```

```python
print("\nCluster Centers:")
```

```python
print(
    pd.DataFrame(
        centers,
        columns=X.columns
    )
)
```

```python
new_signal = [[

    -55,   # RSSI

    -85,   # RSRP

    -9,    # RSRQ

    23,    # SNR

    20,    # Latency

    3      # Packet Loss

]]
```

```python
new_scaled = scaler.transform(
    new_signal
)
```

```python
cluster = kmeans.predict(
    new_scaled
)
```

```python
print("\nNew Signal Cluster:")
print(cluster[0])
print(
    "Signal Quality:",
    quality[cluster[0]]
)
```

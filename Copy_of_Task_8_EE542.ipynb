```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from sklearn.decomposition import PCA

from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
file = '/content/drive/MyDrive/B.TECH/5th Sem/MDM Dataset/wireless_signal_kmeans_dataset(T7,8).csv'
data = pd.read_csv(file)
```

```python
if len(data.columns) == 1:

    print("Single column detected. Fixing CSV format...")

    data = data.iloc[:,0].str.split(
        ",",
        expand=True
    )
```

```python
data.columns = [
        "RSSI",
        "RSRP",
        "RSRQ",
        "SNR",
        "Latency",
        "Packet_Loss"
    ]
```

```python
print("\nDataset Preview:")
print(data.head())
```

```python
print("\nDataset Columns:")
print(data.columns)
```

```python
for col in data.columns:

    data[col] = pd.to_numeric(
        data[col],
        errors="coerce"
    )
```

```python
data.dropna(inplace=True)
```

```python
print("\nMissing Values:")
print(data.isnull().sum())
```

```python
X = data[
    [
        "RSSI",
        "RSRP",
        "RSRQ",
        "SNR",
        "Latency",
        "Packet_Loss"
    ]
]
```

```python
scaler = StandardScaler()
```

```python
X_scaled = scaler.fit_transform(X)
```

```python
plt.figure(figsize=(10,5))
```

```python
linked = linkage(
    X_scaled,
    method="ward"
)
```

```python
dendrogram(linked)
```

```python
plt.title(
    "Hierarchical Clustering Dendrogram"
)
```

```python
plt.xlabel(
    "Samples"
)
```

```python
plt.ylabel(
    "Distance"
)
```

```python
plt.show()
```

```python
model = AgglomerativeClustering(

    n_clusters=3,

    linkage="ward"

)
```

```python
data["Cluster"] = model.fit_predict(
    X_scaled
)
```

```python
print("\nCluster Count:")
```

```python
print(
    data["Cluster"].value_counts()
)
```

```python
cluster_rssi = data.groupby(
    "Cluster"
)["RSSI"].mean()
```

```python
print("\nAverage RSSI:")
print(cluster_rssi)
```

```python
sorted_clusters = cluster_rssi.sort_values(
    ascending=False
)
```

```python
quality = {

    sorted_clusters.index[0] : "Good",

    sorted_clusters.index[1] : "Medium",

    sorted_clusters.index[2] : "Poor"

}
```

```python
data["Signal_Quality"] = data["Cluster"].map(
    quality
)
```

```python
print("\nFinal Classification:")
```

```python
print(

    data[
        [
            "RSSI",
            "RSRP",
            "SNR",
            "Cluster",
            "Signal_Quality"
        ]
    ]

)
```

```python
pca = PCA(
    n_components=2
)
```

```python
X_pca = pca.fit_transform(
    X_scaled
)
```

```python
plt.figure(figsize=(8,6))
```

```python
sns.scatterplot(

    x=X_pca[:,0],

    y=X_pca[:,1],

    hue=data["Signal_Quality"],

    palette="Set1",

    s=100

)
```

```python
plt.title(
    "Hierarchical Clustering of Wireless Signal Quality"
)
```

```python
plt.xlabel(
    "Principal Component 1"
)
```

```python
plt.ylabel(
    "Principal Component 2"
)
```

```python
plt.show()
```
